```python
import pandas as pd
import re
import html
```

```python
comments_df = pd.read_csv("D:/Social media Assignment 2/comments.csv")

comments_df.head()
```

```python
print("Original dataset shape:", comments_df.shape)

comments_df.info()
```

```python
def clean_comment(text):
    if pd.isna(text):
        return None

    text = str(text)

    # Convert HTML entities like &amp; to &
    text = html.unescape(text)

    # Remove URLs only
    text = re.sub(r"http\S+|www\S+|https\S+", "", text)

    # Remove HTML tags
    text = re.sub(r"<.*?>", "", text)

    # Lowercase text
    text = text.lower()

    # Remove excessive whitespace
    text = re.sub(r"\s+", " ", text).strip()

    return text


def is_invalid_comment(text):
    if pd.isna(text):
        return True

    text = str(text).strip().lower()

    invalid_values = [
        "",
        "[deleted]",
        "[removed]",
        "deleted",
        "removed",
        "nan",
        "none"
    ]

    return text in invalid_values
```

```python
comments_cleaned = comments_df[
    ~comments_df["comment_text"].apply(is_invalid_comment)
].copy()

comments_cleaned = comments_cleaned.drop_duplicates(
    subset=["video_id", "author", "comment_text"]
)

print("Shape after removing invalid/duplicate comments:", comments_cleaned.shape)
```

```python
comments_cleaned["clean_comment"] = comments_cleaned["comment_text"].apply(clean_comment)

comments_cleaned = comments_cleaned[
    ~comments_cleaned["clean_comment"].apply(is_invalid_comment)
].copy()

comments_cleaned.head()
```

```python
comments_cleaned["comment_length_chars"] = comments_cleaned["clean_comment"].apply(len)

comments_cleaned["comment_length_words"] = comments_cleaned["clean_comment"].apply(
    lambda x: len(str(x).split())
)

comments_cleaned.head()
```

```python
for comment in comments_cleaned["clean_comment"].head(10):
    print(comment)
    print("-" * 80)
```

```python
comments_cleaned.to_csv("comments_cleaned.csv", index=False, encoding="utf-8")

print("Saved as comments_cleaned.csv")
print("Final cleaned dataset shape:", comments_cleaned.shape)
```
